```python
import json
import numpy as np
from sklearn.model_selection import StratifiedKFold
import pandas as pd
import hashlib
```

```python
seed = 90051
# Use the subset dataset which created by data_preparation.ipynb
csv_path = "expw_subset_metadata_comp90051.csv"
df = pd.read_csv(csv_path)
labels = df["correct_facial_class"].to_numpy()
```

```python
#check the first and last rows
print("Total samples:", len(df),
      "\nFirst 5 rows:",df.head(),
      "\nLast 5 rows:",df.tail())
```

```python
#calculate the SHA value
with open(csv_path, "rb") as f:
    file_hash = hashlib.sha256(f.read()).hexdigest()
print("SHA256:", file_hash)
```

```python
# Outer 10-fold CV
outer_cv = StratifiedKFold(n_splits=10,shuffle=True,random_state=seed)
cv_splits = []
for fold, (train_idx, test_idx) in enumerate(outer_cv.split(np.zeros(len(df)), labels),start=1):
    cv_splits.append({"fold": fold,"train_indices": train_idx.tolist(),"test_indices": test_idx.tolist()})
```

```python
# Save the 10 splits into json file
with open("Outer_10-fold_CV_splits.json", "w") as f:
    json.dump(cv_splits, f, indent=2)
print(f"Saved {len(cv_splits)} outer CV folds",
      f"\nTotal samples: {len(df)}")
```
